```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
cap = cv2.VideoCapture("../images/result_equalize.gif")

while True:
    ret, frame = cap.read()
    if not ret:
        break
    size = frame.shape
    cv2.imshow("Frame", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
    
print(np.unique(size))
cap.release()
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture("./result_equalize.gif")

while True:
    ret, frame = cap.read()
    if not ret:
        break
    frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    frame = frame[0: frame.shape[0], 0: frame.shape[1]//2]

    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(4, 4))
    result_clahe = clahe.apply(frame)

    result_hist = cv2.equalizeHist(frame)

    gray_clahe_hist = np.hstack((frame, result_clahe, result_hist))
    cv2.imshow("Gray _ CLAHE _ equalhist", gray_clahe_hist)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
    
cap.release()
cv2.destroyAllWindows()
```

```python
# gamma correction

cap = cv2.VideoCapture("./result_equalize.gif")

while True:
    ret, frame = cap.read()
    if not ret:
        break
    frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    frame = frame[0: frame.shape[0], 0: frame.shape[1]//2]

    gamma = 1.2
    lookUpTable = np.empty((1,256), np.uint8)
    for i in range(256):
        lookUpTable[0,i] = np.clip(pow(i / 255.0, gamma) * 255.0, 0, 255)
    result_gamma = cv2.LUT(frame, lookUpTable)

    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(4, 4))
    result_clahe = clahe.apply(result_gamma)

    result_hist = cv2.equalizeHist(result_gamma)

    gray_clahe_hist = np.hstack((frame,result_gamma, result_clahe, result_hist))
    cv2.imshow("Gray _Gamma Correction _ CLAHE _ equalhist", gray_clahe_hist)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
    
cap.release()
cv2.destroyAllWindows()
```
