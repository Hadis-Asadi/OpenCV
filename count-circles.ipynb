```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("images/blobs (1).jpg")

img_copy = img.copy()
img_copy1 = img.copy()

params = cv2.SimpleBlobDetector_Params()

params.filterByInertia = True
params.minInertiaRatio = 0.5
detector = cv2.SimpleBlobDetector_create(params)
keypoints = detector.detect(img)
result = cv2.drawKeypoints(img_copy, keypoints, np.array([]), (0,0,255), cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

params_non_circle = cv2.SimpleBlobDetector_Params()
params_non_circle.filterByInertia = True
params_non_circle.maxInertiaRatio = 0.5
detector_non_circle = cv2.SimpleBlobDetector_create(params_non_circle)
keypoints_non_circle = detector_non_circle.detect(img)

result_non_circle = cv2.drawKeypoints(img_copy1, keypoints_non_circle, np.array([]), (0,255), cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

print("Number of circles: ", len(keypoints))
print("Number of non-circles: ", len(keypoints_non_circle))

plt.figure(figsize=(15, 10))
plt.subplot(121)
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title("circles")

plt.subplot(122)
plt.imshow(cv2.cvtColor(result_non_circle, cv2.COLOR_BGR2RGB))
plt.title("non-circles")
```
