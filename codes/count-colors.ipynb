```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
hue_spectrum = np.linspace(0, 179, 180).astype(np.uint8 )

img = np.array([hue_spectrum]* 30)

saturation = 255 * np.ones_like(img)
value = 255 * np.ones_like(img)

final_img = cv2.merge((img, saturation, value))
final_img = cv2.cvtColor(final_img, cv2.COLOR_HSV2BGR)      
plt.imshow(final_img[...,::-1])
```

```python
input_img = cv2.imread("../images/result_bubbles_colored.png.jpeg")

img_hsv = cv2.cvtColor(input_img, cv2.COLOR_BGR2HSV)

plt.figure(figsize=(10, 10))
plt.imshow(input_img[...,::-1])
```

```python
lower_b = np.array([90,20,20])
upper_b = np.array([130,255,255])
mask_b = cv2.inRange(img_hsv, lower_b, upper_b)
result_b = cv2.bitwise_and(input_img, input_img, mask=mask_b) 
contours_b, _ = cv2.findContours(mask_b, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)  
all_b = cv2.drawContours(result_b.copy(), contours_b, -1, (0,0,255), 2)
print("counts of blue=", len(contours_b))

plt.figure(figsize=(15,15))
plt.subplot(131);plt.imshow(result_b[...,::-1]);plt.title("Original");
plt.subplot(132);plt.imshow(mask_b, cmap='gray');plt.title("Mask");
plt.subplot(133);plt.imshow(all_b[...,::-1]);plt.title("Contours");
```

```python
# use a dictionary to optimize the code and avoid repetition
color_ranges = {
    "blue": ([90,20,20], [130,255,255]),
    "green": ([35,50,50], [80,255,255]),
    "yellow": ([30,50,50], [35,255,255]),
    "orange": ([15,50,50], [25,255,255]),
    "purple": ([125,5,5], [155,255,255])
}
for color, (lower, upper) in color_ranges.items():
    mask = cv2.inRange(img_hsv, np.array(lower), np.array(upper))
    result = cv2.bitwise_and(input_img, input_img, mask=mask) 
    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)  
    all_contours = cv2.drawContours(result.copy(), contours, -1, (0,0,255), 2)
    print(f"counts of {color}=", len(contours))

    plt.figure(figsize=(15,15))
    plt.subplot(131);plt.imshow(result[...,::-1]);plt.title("Original");
    plt.subplot(132);plt.imshow(mask, cmap='gray');plt.title("Mask");
    plt.subplot(133);plt.imshow(all_contours[...,::-1]);plt.title("Contours");
```
