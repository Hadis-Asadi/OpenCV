```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = cv2.imread("images/result_swap.png.jpeg")
plt.imshow(img[...,::-1])
h,w,_= img.shape
```

```python
img1 = img[0:h//2, 0:w//2]
img2 = img[h//2:h, 0:w//2]
img3 = img[0:h//2, w//2:w]
img4 = img[h//2:h, w//2:w]

plt.figure(figsize=(5,5))
plt.subplot(221);plt.title("Image 1");plt.imshow(img1[...,::-1]);plt.axis('off')
plt.subplot(222);plt.title("Image 2");plt.imshow(img2[...,::-1]);plt.axis('off')
plt.subplot(223);plt.title("Image 3");plt.imshow(img3[...,::-1]);plt.axis('off')
plt.subplot(224 );plt.title("Image 4");plt.imshow(img4[...,::-1]);plt.axis('off')
```

```python
zeros = np.zeros_like(img)

zeros[0:h//2, 0:w//2] = img4
zeros[h//2:h, 0:w//2] = img3
zeros[0:h//2, w//2:w] = img2
zeros[h//2:h, w//2:w] = img1
plt.imshow(zeros[...,::-1])
plt.title("correct swap");plt.axis('off')
```
